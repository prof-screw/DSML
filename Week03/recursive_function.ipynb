```python
cache= {0: 0, 1:1}

def fib(n):
    
    if n in cache:
        return cache[n]

    # if n==0:
    #     return 0
    # elif n==1:
    #     return 1
    else:
        res= fib(n-1) + fib(n-2)
        cache[n]=res
        return res
    
        
```

```python
%%time

fib(1000)
```

```python
def fact(num):
    if num==0:
        return 1
    return num*fact(num-1)

fact(6)
```

```python
cache_fact= [1,1]

def factorial(n):
    if n< len(cache_fact):
        return cache_fact[n]
    else:
        res= n*factorial(n-1)
        cache_fact.append(res)
        return res
    
factorial(5)
```
