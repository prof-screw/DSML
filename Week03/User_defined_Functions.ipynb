```python
def sum_of_all(*args):   # '*' is a positional argument args is a list that stores in order of what it is
    return sum(args)

result= sum_of_all(1,234,45,65,64,356,43,7,4,63,456,45)
print(result)
```

```python
def print_kwargs(**kwargs):
    print("Keyword Arguments") #stores in dictionary like how positional stores in list
    print(kwargs)
    
print_kwargs(num1=1, bool1= True)
# print_kwargs(1,2) this will throw error
```

```python
def calculate_area(shape: str, **kwargs):
    area=0
    if shape == "square":
        area= kwargs.get("length", 0)** 2
    elif shape =="rectangle":
        if "length" in kwargs.keys() and "breadth" in kwargs.keys():
            area=kwargs.get("length", 0)* kwargs.get("breadth", 0)
    else:
        print(f"{shape} is not supported")
        
    return area

print(calculate_area("rectangle", length=5, breadth=10))   
```
