```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, length,breadth=0, height=0):
        self.l= length
        self.b= breadth
        self.h= height
    
    @abstractmethod
    def area(self):
        pass
    
    def volume(self):
        return "Not applicable for given shape"
        
```

```python
class Rectangle(Shape):
    def area(self):
        return self.l* self.b
    
    
    
```

```python
rect= Rectangle(50,40)
rect.area()
```

```python
import math

class Circle(Shape):
    def __init__(self, radius):
        self.r= radius
        
    def area(self):
        return math.pi * self.r ** 2
```

```python
cir= Circle(5)
cir.area()
```
