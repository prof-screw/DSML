```python
def get_cube(num):
    return num**3
```

```python
ls= [1,4,8,2,6,9]
```

```python
map_obj= map(get_cube, ls)
for val in map_obj:
    print(val)
```

```python
list(map_obj) #this will return an empty list because the map object has already been exhausted in the previous for loop.

```

```python
filter_obj= filter(lambda num: num%2==0, ls) #filter statement checks if the condition is true in a given iterable and then returns the values that's true
# filter_obj= filter(get_cube, ls) #this returns all values because get cube function returns all non-zero values, which is true
```
