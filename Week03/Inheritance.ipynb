```python
class Animal:
    
    def __init__(self, name, age):
        self.name= name
        self.age= age
        
    def intro(self):
        print(f"Hello, i am {self.name}")
    
```

```python
class cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name,age)
        self.color= color
        
    def details(self):
        print(f"{self.color} cat: \n{self.name} \n age: {self.age}")
        
child1=cat("kitty", 15, "Orange")
child1.details()
```

```python
# Designing a simple hierarchy of employees in a company
 
#     - The Person class stores basic details like name and age.
 
#     - The Employee class inherits from Person and adds an employee ID.
 
#     - The Manager class inherits from Employee and adds a department.
 
# Write these classes and implement a method get_details() in the Manager class that returns a string with name, age, employee ID, and department.
 
 
class Person:
    def __init__(self,name:str, age:int):
        self.name=name
        self.age=age

class Employee(Person):
    def __init__(self,name:str, age:int, emp_id:int):
            super().__init__(name,age)
            self.empy_id=emp_id
        
    
class Manager(Employee):
    def __init__(
        self,
        name:str,
        age: int,
        emp_id:int,
        department: str
    ):
        super().__init__(name, age, emp_id)
        self.dept= department
    
    def get_details(self)-> str:
        return f"Name: {self.name}, Age: {self.age}, Employee ID: {self.empy_id}, Department: {self.dept}"
    
m = Manager("Superman", 35, 101, "CSIT")
print(m.get_details())
        
```
