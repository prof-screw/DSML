```python
class Calculator:
    
    def __init__(self, num1, num2, *args):
        print("Making objects...")
        self.num1= num1
        self.num2= num2
        pass
    
    def add(self):
        return self.num1 + self.num2
    
    def multiply(self, num3: int)-> int | float:
        return self.num1 * self.num2 * num3
```

```python
calc= Calculator(5,9)
print(calc.multiply(10))
```

```python
print(calc.num1+calc.num2)
```
