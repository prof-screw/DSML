```python
class Calculator:
    
    def __init__(self, num1, num2, *args):
        print("Making objects...")
        self.num1= num1
        self.num2= num2
        pass
    
    def add(self):
        return self.num1 + self.num2
    
    def multiply(self, num3: int)-> int | float:
        return self.num1 * self.num2 * num3
```

```python
calc= Calculator(5,9)
print(calc.multiply(10))
```

```python
print(calc.num1+calc.num2)
```

```python
class Person:
    count=0
    
    def __init__(self, name:str, address: str= "Kathmandu"):
        #initializer or constructor
        self.name= name  #instance attribute
        self.address= address  #instance attribute
        
        Person.count +=1
        
    def get_details(self):  #instance method
        print(f"I am {self.name} from {self.address}")
    
    @classmethod    
    def get_count(cls): 
        return cls.count
        
```

```python
student1= Person("Brandon", "norway")
student1.get_details()
student1.get_count()
```
