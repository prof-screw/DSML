```python
def sum_natural ( num):
    sum=0
    for i in range(num+1):
        sum += i
    return sum
print(sum_natural(5))
```

```python
#type hinting
def sumy (num1:int, num2:int)->int: #basically this in C==> int sumy (int num1, int num2){}
    """_sum numbers_

    Args:
        num1 (int): _first number_
        num2 (int): second number

    Returns:
        int: sum of 2 values
    """
    return num1+num2
print(sumy(1,2))
```
