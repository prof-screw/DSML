```python
# Write a Program that Uses Functions write_to_file and read_from_file:

#     write_to_file(filename, content): Writes content to a file named filename. If the file doesn't exist, it should be created.
#     read_from_file(filename): Reads and prints the content of a file named filename. Call write_to_file to write "Hello, Python!" to a file named "greetings.txt", then call read_from_file to read and print the content of this file.


def write_to_file(filename:str, content:str):
    with open(filename, "w") as file_obj:
        file_obj.write(content)
    print("Message written on text file.\n")

write_to_file("hello.txt", "This is the message")

def read_from_file(filename:str)->str:
    with open(filename, "r") as reader_obj:
        message= reader_obj.read()
    return message

message_txt= read_from_file("hello.txt")
print(message_txt)
        
```

```python
# Task 01: Create a Book Filtering Function

# Given the list books as shown below, write a Python function named filter_books that filters books based on genre and publication year. The function should take two parameters: genre (a string) and year (an integer). 
# It should return a list of book titles that match the given genre and have been published on or after the specified year.

#     Example usage : print(filter_books("Fiction", 1980))
#     Expected output: ['The Alchemist', 'The Catcher in the Rye']

# Try to use List Comprehension with If condition

def filter_books(genre: str, year: int)-> list:
    filtered_list = [row[0] for row in books if row[1]==genre and row[2]>=year]
    return filtered_list

print(filter_books("Fiction", 1980))

# def filter_books(genre: str, year: int)-> list:
#     filtered_list = [title  for title, chk_genre, chk_year, _ in books if chk_genre==genre and chk_year>=year]
#     return filtered_list

# print(filter_books("Fiction", 1980))
```
