```python
# Write a Program that Uses Functions write_to_file and read_from_file:

#     write_to_file(filename, content): Writes content to a file named filename. If the file doesn't exist, it should be created.
#     read_from_file(filename): Reads and prints the content of a file named filename. Call write_to_file to write "Hello, Python!" to a file named "greetings.txt", then call read_from_file to read and print the content of this file.


def write_to_file(filename:str, content:str):
    with open(filename, "w") as file_obj:
        file_obj.write(content)
    print("Message written on text file.\n")

write_to_file("hello.txt", "This is the message")

def read_from_file(filename:str)->str:
    with open(filename, "r") as reader_obj:
        message= reader_obj.read()
    return message

message_txt= read_from_file("hello.txt")
print(message_txt)
        
```

```python

# Control Flow with Nested Loops and Complex Logic

# Write a Python program that simulates a number guessing game:

#     The program should generate a random number between 1 and 100 and give the user 7 attempts to guess it.

#     After each wrong guess, the program should provide a hint whether the guess was too high or too low.

#     If the user fails to guess the number within the attempts, the program should reveal the number and ask if they would like to play again.


import random
random_number = random.randint(1, 100) 

# Your Solution Here
user_response="y"
while user_response=="y":
    print("Guess the Number between 0-100\n")
    
    for i in range(7):
               
        guess=int(input())
        guess_correct= False
        if guess==random_number:
            print("You Guessed Correct!\n")
            guess_correct=True
            break
        
        difference = abs(random_number - guess)
        if difference<=5:
            print("You're almost there!")
        elif difference <= 10:
            print("You're within range!")
        elif difference <= 25:
            print("You're getting close\n")
        elif difference <= 50:
            print("You're close to halfway \n")
        elif difference <= 75:
            print("You're quite far\n")
        else:
            print("You're way off\n")
        
        if guess > random_number: 
            print("Your guess is too high.\n")
        else: 
            print("Your guess is too low.\n")        
            
    if guess_correct == False:
        print("Failed Guess.") 
        print("The number was:", random_number)
        
    print("Failed Guess. Wanna try again? y/n")
    user_response= str(input())
    if user_response== "n":
        print("Thanks for playing\n")
        break

```

```python
# Question:

# Write a Python program to check whether a given number is an Armstrong number or not.

# Definition: An Armstrong number (also known as a narcissistic number) is a number that is equal to the sum of its own digits each raised to the power of the number of digits. For example:

#     153 is an Armstrong number because ( 1^3 + 5^3 + 3^3 = 153 ).
#     9474 is an Armstrong number because ( 9^4 + 4^4 + 7^4 + 4^4 = 9474 ).

# Input:
# An integer (e.g., 153).

# Output:
# Output "Yes, it's an Armstrong number." if the number is an Armstrong number. Otherwise, output "No, it's not an Armstrong number."

# Constraints:

#     The input should be a positive integer.

def is_armstrong_number(num):
    digits = str(num)
    power = len(digits)
    total = sum(int(d) ** power for d in digits)
    return total == num

num = int(input("Enter a positive integer: "))
if is_armstrong_number(num):
    print("Yes, it's an Armstrong number.")
else:
    print("No, it's not an Armstrong number.")

```

```python
# Write a Python program that iterates through integers from 1 to 50. For each multiple of three, print "Fizz" instead of the number; for each multiple of five, print "Buzz". For numbers that are multiples of both three and five, print "FizzBuzz".

# The FizzBuzz problem is a common coding challenge that is often used in programming interviews to test basic programming skills. The problem typically requires writing a function that prints numbers from 1 to a given limit, but with a twist:

#     For multiples of 3, print "Fizz" instead of the number.
#     For multiples of 5, print "Buzz" instead of the number.
#     For numbers which are multiples of both 3 and 5, print "FizzBuzz".

for i in range(1,50):
    skip_flag=False
    if i%3==0 and i%5==0:
        print("FizzBuzz")
        skip_flag=True
    elif i%3==0 and i%5!=0:
        print("Fizz")
        skip_flag=True
    elif i%3!=0 and i%5==0:
        print("Buzz")
        skip_flag=True
    if skip_flag==True:
        continue
    else:
        print(i)

    
```
