```python
#r+ first open file for reading and writing. if file does not exist, it will throw an error
#w+ first open file for writing and reading. if file does not exist, it will create a new file
#a+ first open file for appending and reading. if file does not exist, it will create a new file
```

```python
file_obj= open("msg.txt", "w")
file_obj.write("hello world\n")
file_obj.write("hello back\n")
file_obj.close()
```

```python
with open("msg.txt", "r+") as file_obj:
    message= file_obj.read()
    print(message)
    file_obj.write("\nhello from the other side\n")
    print(file_obj.read()) #this will return nothing because the file pointer is at the end of the file after writing. so, it will read nothing from the file
print("closed file")
```

```python
with open("msg.txt", "r") as file_obj:
    msg_lines= file_obj.readlines()
    print(msg_lines) #returns a list of lines in the file. each line is an element in the list
```
