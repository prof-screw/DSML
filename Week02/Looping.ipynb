```python
fruits= ["apple", "banana", "cherry", "date", "elderberry"]
for fruit in fruits:
    if fruit == "cherry":
        print(fruit)
```

```python
adjectives = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for ft in fruits:
    for adj in adjectives:
        print(adj, ft)
```
