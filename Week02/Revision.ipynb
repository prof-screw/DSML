```python
# Python
# Input matrix:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Expected Output: [4, 16, 36, 64]
# Your code here:
squared_evens = [x**2 for mat in matrix for x in mat if x % 2 == 0]
print(squared_evens)

```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
squared_evens=[]
for items in matrix:
    for num in items:
        if num % 2 == 0:
            squared_evens.append(num**2)

print(squared_evens)           
```
