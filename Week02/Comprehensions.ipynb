```python
#dictionary comprehension
squared_dict= {i: i**2 for i in range(10) if i%2==0} #i is key and after semicolon (;)the result is the value of the key.
squared_dict
```

```python
#  Count length of words ( if length > 4)
sent = "Although she was extremely tired after working a long shift at the local hospital, Maria still managed to stop by the grocery store, pick up fresh vegetables for dinner, and help her younger brother with his math homework."
 
# O/P : {words : len(word) } , if length of word is > 4
words= sent.split()
words_dict= {words: len(words) for words in words if len(words)>4}
print(words_dict )  

res={}
for words in sent.split():
    if len(words)>4:
        res[words]= len(words)
print(res)
```
