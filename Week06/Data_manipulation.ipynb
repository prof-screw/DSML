### Handling null or NaN or NA values

```python
titanic_df.isna().sum()
```

1. Removing Null values
        only when a few information is lost (like embarked). cant remove something like Age or Cabin row data

2. Fill something (imputation)


3. Nul as new feature (Feature engineering)

0-> column wise

```python
titanic_df.dropna(subset= "Embarked", axis=0, ignore_index=True, inplace=True)
```

```python
titanic_df.shape
```

```python
#null value as feature
titanic_df.Cabin=titanic_df.Cabin.notna().astype(int)
```

```python
titanic_df.isna().sum()
```

```python
titanic_df.head(5)
```

```python
# Fill Null values
titanic_df_fillAge = titanic_df.copy()

titanic_df_fillAge.loc[:, "Age"] = titanic_df_fillAge["Age"].fillna(
    value=titanic_df["Age"].median()
)

titanic_df_fillAge.isna().sum()
```

```python
# Inputation of age wrt initials
group_by_initial= titanic_df.groupby(by= "Initials")
```

```python
Median_age= group_by_initial["Age"].transform("median")
```

```python
Median_age
```

```python
titanic_df_fillAge_with_initial= titanic_df.copy()

titanic_df_fillAge_with_initial.loc[:, "Age"]= titanic_df_fillAge_with_initial["Age"].fillna(Median_age)
```

```python
titanic_df_fillAge_with_initial.Age.plot
```

```python
titanic_df.Initials = titanic_df.Initials.replace(to_replace=["Rev", "Col", "Major", "Capt", "Jonkheer", "Don",  "Sir"]
                            , value="Mr").value_counts()
```

```python
titanic_df.Initials =titanic_df.Initials.replace({"Mlle": "Miss", "Ms": "Miss", "Mme": "Mrs", "Countess": "Mrs", "Lady":"Mrs"})
```
