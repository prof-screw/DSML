```python
# %pip install seaborn
import pandas as pd
titanic_df= pd.read_csv("titanic_train.csv")
```

1. Bar Graph


```python
titanic_df.value_counts(subset= ["Sex", "Survived"]).plot(
    kind="bar",
    title="Count of Suvived Passengers",
    figsize=(4,3),
    legend=True,
    xlabel= "survived (0=no, 1=yes)",
    ylabel="no. of passengers",
    grid=False,
    yticks=range(0,600,50),
    colormap="Pastel2"
    )
```

```python
ax= pd.crosstab(
titanic_df["Sex"], titanic_df["Survived"]
).plot(
    kind= "bar"
)
ax.bar_label(ax.containers[0],padding=2)
# titanic_df.value_counts(subset= ["Sex", "Survived"])in.plot(
#     kind="bar",
#     title="Count of Suvived Passengers",
#     figsize=(4,4),
#     legend=True,
#     xlabel= "survived (0=no, 1=yes)",
#     ylabel="no. of passengers",
#     grid=False,
#     yticks=range(0,600,50),
#     colormap="Pastel2"
#     )
```

Pie chart


```python
group_by_embarked = titanic_df.groupby(by="Embarked")
group_by_embarked.count()
```

```python
group_by_embarked.count().plot(
    kind= "pie",
    y= "Survived",
    title= "Count of Passengers according to Pclass",
    autopct= "%.2f%%", #syntax %"places of integer"."decimal placed how many"f%final%to show "%"
    startangle=90
)   
```

```python
crosstab= pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
)
crosstab
```

```python
crosstab.T.plot(
    kind="pie",
    subplots= True,
    legend=True
)
```

7. Heatmap

```python
correlation= titanic_df.corr(numeric_only=True)
correlation #.style.background_gradient(cmap="coolwarm")
```

```python

import seaborn as sns
```

```python
sns.heatmap(
    data=correlation,
    cmap= "coolwarm",
    fmt= ".2f",
    annot=True)
```

```python
import numpy as np

mask= np.triu(np.ones_like(correlation,  dtype=bool))
```

```python
sns.heatmap(data= correlation, cmap="coolwarm", fmt= ".2f", annot=True, mask =mask)
```
