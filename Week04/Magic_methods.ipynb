```python
class Person:
    """
    hello
    """
    def __init__(self,name,age):
        self.name= name
        self.age=age
     
    def __str__(self):
        return f"{self.name} / {self.age}"
     
    def __repr__(self):
         return f"Person {self.name}, {self.age}" 
     
    def __len__(self):
        return self.age
    
    def __call__(self, gender= "male"):
        return f"{self.name} is {self.gender}"
```

```python
person = Person("hamilton", 32)
```

```python
print(person)
```
