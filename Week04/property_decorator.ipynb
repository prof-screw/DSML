```python
#setter method
class Person:
    def __init__(self, name):
        self.__full_name= name
        
    @property
    def name(self):   #getter method
        print("Getting name")
        return self.__full_name
    
    @name.setter #setter method
    def name(self, value):
        print("setting name: ")
        self.__full_name = value
        
    @name.deleter
    def name(self):
        print("Deleting name")
        self.__full_name="" #we can implement our own version of 'delete' either deleting obj or just assigning null value to it
```

```python
person= Person("Jack sparrow")
```

```python
person.name
```

```python
person.name= "Edward scissorhands"
```

```python
person.name
```

```python
del person.name
```
