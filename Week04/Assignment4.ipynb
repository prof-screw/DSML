Math Tutor Using Random, Arithmetic Operators, and OOP

Scenario: Create a MathTutor class that generates random math questions using random and math operators.

    Track correct and incorrect answers.
    Provide a score at the end.
    Handle invalid input using exception handling.

    Concepts: Random, Math, Exception Handling, Classes


```python
import random
from ModulesAndPackages.calculator.calculator_pkg import add, subtract, multiply, divide


class MathTutor:

    operations = {"+": add, "-": subtract, "*": multiply, "/": divide}
    op_symbols = list(operations.keys())

    def __init__(self, num_questions=5):
        self.num_questions = num_questions
        self.correct = 0
        self.incorrect = 0

    def question_generator(self):
        num1 = random.randint(-1000, 1000)
        num2 = random.randint(-1000, 1000)
        op_symbol = random.choice(self.op_symbols)

        # never let division question have a 0 num2
        while op_symbol == "/" and num2 == 0:
            num2 = random.randint(-1000, 1000)

        try:
            correct_answer = self.operations[op_symbol](num1, num2)
        except ZeroDivisionError:
            #division module already has an exception handling but this is just in case it doesnt
            print("Skipping a division-by-zero question.\n")
            return

        user_answer = self._get_valid_answer(num1, op_symbol, num2)
        self.score_tally(user_answer, correct_answer)

    def _get_valid_answer(self, num1, op_symbol, num2):
        """Keep prompting until the user gives a valid number."""
        while True:
            raw = input(f"{num1} {op_symbol} {num2} = ")
            try:
                return float(raw)
            except ValueError:
                print("That's not a valid number. Please try again.")

    def score_tally(self, user_answer, correct_answer):
        # float comparison with round off at 2 decimal places
        if round(user_answer, 2) == correct_answer:
            self.correct += 1
            print("Correct!\n")
        else:
            self.incorrect += 1
            print(f"Incorrect. The correct answer was {round(correct_answer,2)}.\n")

    def run(self):
        for _ in range(self.num_questions):
            self.question_generator()
        self.score_printer()

    def score_printer(self):
        total = self.correct + self.incorrect
        print(f"Final Score: {self.correct}/{total} correct")

if __name__ == "__main__":#makes calculator pkg import it only work in assignmet4.ipynb and not importable
    try:
        tutor = MathTutor(5)
        tutor.run()
    except KeyboardInterrupt:
        print("\nSession ended early.")
```
