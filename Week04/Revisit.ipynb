```python
#prime number

# filter prime numbers from the given list
test_numbers = [

    -10, -1, 0, 1, 2, 3, 4, 5, 6, 7, 9, 11, 13, 15, 17, 19, 21, 25, 27, 47, 49, 79, 97, 100, 523, 527, 1000, 7919, 10000, 104729, 1000000

]


def isPrime(numList):
    prime_list = []
    for i in numList:
        if i < 2: 
            continue
        is_prime = True
        for j in range(2, int(i ** 0.5) + 1):
            if i % j == 0:
                is_prime = False
                break 
        if is_prime:
            prime_list.append(i)

    return prime_list


print(isPrime(test_numbers))        
    
```
