```python
def decorator(func):
    
    def wrapper():
        
        print("Before funciton call")
        print(func().upper())
        print("After function call")
        
    return wrapper
```

```python
@decorator
def greet():
    return "Hellow, world"
```

```python
import time

def timeit(func):
    """total time taken by function run

    Args:
        func (_type_): _description_
    """
    
    def inner(user_email:str):
        start= time.time()
        res= func(user_email)
        
        print("total time taken", int(start- time.time()), "seconds")
        return res
    
    return inner
```
