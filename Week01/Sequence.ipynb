```python
fruits= ['apple', 'banana', 'cherry', 'kiwi', 'mango']
fruits.append('orange')  # adds an element to the end of the list
fruits.insert(1, 'blueberry')  # adds an element at a specific index
fruits.remove('kiwi')  # removes the first occurrence of an element
fruits.extend(['grape', 'pear'])  # adds multiple elements to the end of the list
fruits[2]= 'strawberry'  # modifies an element at a specific index
x=fruits+['pineapple', 'watermelon']  # creates a new list by concatenating two lists
print(fruits)
print(x)
```

```python
print(x)
# x.remove('cherry')  # removes the first occurrence of an element
x.pop(2)  # removes the element at index 2
```

```python
print(x)
del x[2]
print(x)
x.clear()
print(x)
```

```python
y= fruits.copy()  # creates a shallow copy of the list
z= fruits #creates a reference to the original list
fruits.append('papaya')
print(y)
print(z)
```
