```python
import pandas as pd

titanic_df= pd.read_csv("clean_titanic_train.csv")

titanic_df.head(5)
```

### Correlation
we can remove strong correlated columns

```python
# titanic_df.drop("Unnamed: 0", axis=1, inplace=True)
titanic_df.corr(numeric_only=True)
```

```python
import scipy.stats as stats
```

```python
contengency_table= pd.crosstab(titanic_df.Pclass, titanic_df.Survived)
contengency_table
```

```python
chi2_score, pvalue, dof, expected_frequency= stats.chi2_contingency(contengency_table)
```

```python
print(f"Chi-Squared test score= {round(chi2_score, 3)}")
print(f"p-value= {round(pvalue, 5)}")
```

```python
contingency_table= pd.crosstab(titanic_df.Cabin, titanic_df.Parch)
```

```python
chi2_score, pvalue, dof,_= stats.chi2_contingency(contingency_table)
```

```python
print(f"Chi-Squared test score= {round(chi2_score, 3)}")
print(f"p-value= {round(pvalue, 5)}")
```
