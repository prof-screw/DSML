```python
# Path of the file to read
flight_filepath = "../Dataset/flight_delays.csv"

# Read the file into a variable flight_data
flight_data = pd.read_csv(flight_filepath, index_col="Month")
```

```python
flight_data.head()
```

```python
plt.figure(figsize=(10,6))

plt.title("average arrival delay for spirit Airlines, by months")

sns.barplot(x=flight_data.index, y=flight_data.NK)

plt.ylabel("Arrival delay(in minutes)")
```

```python
plt.figure(figsize=(14,7))

plt.title("average arrival delay for each Airlines, by months")

sns.heatmap(data=flight_data, annot=True)

plt.xlabel("Airline")
```

```python
insurance_filepath = "../Dataset/insurance.csv"

# Read the file into a variable insurance_data
insurance_data = pd.read_csv(insurance_filepath)
```

```python
insurance_data.head()

```

```python
sns.scatterplot(x= insurance_data['bmi'], y=insurance_data['charges'])
```

```python
sns.regplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

```python
sns.scatterplot(x= insurance_data['bmi'], y=insurance_data['charges'], hue=insurance_data['smoker'])
```

```python
# Path of the file to read
iris_filepath = "../Dataset/iris.csv"

# Read the file into a variable iris_data
iris_data = pd.read_csv(iris_filepath, index_col="Id")

# Print the first 5 rows of the data
iris_data.head()

```

```python
sns.histplot(iris_data['Petal Length (cm)'])
```

```python
sns.kdeplot(data=iris_data['Petal Length (cm)'], fill=True)
```

```python
sns.jointplot(x=iris_data['Petal Length (cm)'], y=iris_data['Sepal Width (cm)'], kind="kde")
```

```python
sns.histplot(data=iris_data, x='Petal Length (cm)', hue='Species')

# Add title
plt.title("Histogram of Petal Lengths, by Species")
```

```python
sns.kdeplot(data=iris_data, x='Petal Length (cm)', hue='Species', fill=True)

# Add title
plt.title("Distribution of Petal Lengths, by Species")
```
