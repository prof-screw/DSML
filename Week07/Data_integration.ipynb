```python
import pandas as pd

df1= pd.DataFrame(
    {"acc": [13,21,37,40],
     "name": ["Ram", "shyam", "Hari", "Krishna"],
     "age": [21,65,34,88]}
)
df2= pd.DataFrame(
    {"acc": [13,21,37,40],
     "principle": [500, 658,4032, 500],
     "interest": [6.3,4.25,9.8,5.6]}
)
```

```python
df1, df2
```

```python
pd.concat([df1,df2], axis=1)
```

```python
pd.concat([df1,df2], axis=0, ignore_index=True)
```

```python
df3= pd.DataFrame(
    {"key": ["ab", "cb", "ac","de"],
     "name": ["Ram", "shyam", "Hari", "Krishna"],
     "age": [21,65,34,88]}
)
df4= pd.DataFrame(
    {"key": ["ab","ge","de","fc"],
     "principle": [500, 658,4032, 500],
     "interest": [6.3,4.25,9.8,5.6]}
)
```

```python
pd.merge(df3,df4,how= "outer", on="key")
```

```python
pd.merge(df3,df4,how= "right", on="key")
```

```python
pd.merge(df3,df4,how= "inner", on="key")
```
