Matrix Operation

```python
bT=np.transpose(b)
```

```python
# All possible matrix multi methods
a @ bT #discouraged method
np.dot(a,bT)
res=a.dot(bT) 
```

```python
#Trace
d_sum= np.trace(res)
print(d_sum)
```

```python
#Rank
np.linalg.matrix_rank(a)
```

```python
#determinant
d= np.random.randint(1,45, size= (3,3))
det_d=np.linalg.det(d) #only square matrix can have determinants
```
