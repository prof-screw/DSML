```python
def number_generator(start, end):
    for num in range(start, end+1):
        yield num
    
```

```python
gen= number_generator(10,23)
```

```python
next(gen)
```

```python
my_gen= number_generator(1,6)
for num in my_gen:
    print(num)
    
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i==0:
            return False
    return True
```

```python
def prime_generator(limit=10):
    count, start= 1,1
    
    while count<= limit:
        start +=1
        if is_prime(start):
            count += 1
            yield start
        
    
```

```python
prime_gen= prime_generator(10)
```

```python
for i in prime_gen:
    print(i)
```
