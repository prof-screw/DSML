why it's better than python for loops?
cause it's designe to perform operations on entire arrays rather than individual elements. also enables caching. also also written in C so it's way faster

```python
import time
import numpy as np
```

```python
a=np.random.rand(90_000_000)
b=np.random.rand(90_000_000)
```

```python
st=time.time()
res= []
for num1, num2 in zip(list(a), list(b)):
    res.append(num1*num2)
    
end= time.time()
```

```python
print(f"time taken by python loop= {end-st}")
```

```python
st=time.time()
res= [ num1* num2 for num1, num2 in zip(list(a), list(b)) ]
end=time.time()
```

```python
print(f"time taken by python loop= {end-st}")
```

```python
st=time.time()
res= np.multiply(a,b)
end= time.time()
```

```python
print(f"time taken by numpy: {end-st} sec")
```
