```python
#how many matches ended in a draw, home win, or away win
def get_result(row):
    if row["home_score"] > row["away_score"]:
        return "H"
    elif row["home_score"] < row["away_score"]:
        return "A"
    else:
        return "D"

matches_clean["target"] = matches_clean.apply(get_result, axis=1)
matches_clean["target"].value_counts()
```

```python
teams_of_interest = ["Arsenal", "Manchester United", "Liverpool", "Manchester City", "Leicester City"]
```

```python
home_goals = matches_clean.groupby(["season_start_year", "home"])["home_score"].sum()
home_goals = home_goals.reset_index()
home_goals.columns = ["season_start_year", "team", "goals"]

away_goals = matches_clean.groupby(["season_start_year", "away"])["away_score"].sum()
away_goals = away_goals.reset_index()
away_goals.columns = ["season_start_year", "team", "goals"]

team_goals = pd.concat([home_goals, away_goals])
team_goals = team_goals.groupby(["season_start_year", "team"])["goals"].sum().reset_index()
team_goals.head()
```

```python
subset = team_goals[team_goals["team"].isin(teams_of_interest)]

plt.figure(figsize=(19,10))
ax= plt.gca()
ax.xaxis.set_major_locator(MultipleLocator(1))
plt.xticks(rotation=45)
sns.lineplot(data=subset, x="season_start_year", y="goals", hue="team", marker="o")
plt.title("Goals Scored per Season, Selected Teams")
plt.xlabel("Season Start Year")
plt.ylabel("Total Goals Scored")
plt.show()
```

```python
#League position vs Match outcome
all_tables_renamed = all_tables.rename(columns={"team_full_name": "home", "Year": "season_start_year", "Place": "home_position"})
all_tables_renamed = all_tables_renamed[["home", "season_start_year", "home_position"]]

#merge each match with position of home team end of season position
matches_with_position = matches_clean.merge(all_tables_renamed, on=["home", "season_start_year"], how="left")
matches_with_position[["home", "season_start_year", "home_position"]].head()
```

```python
# Bring in away team's position too
away_table = all_tables.rename(columns={"team_full_name": "away", "Year": "season_start_year", "Place": "away_position"})
away_table = away_table[["away", "season_start_year", "away_position"]]

matches_with_position = matches_with_position.merge(away_table, on=["away", "season_start_year"], how="left")
matches_with_position[["home", "away", "home_position", "away_position", "target"]].head()
```

```python
matches_with_position["position_gap"] = matches_with_position["away_position"] - matches_with_position["home_position"]
print(matches_with_position["home_position"].isnull().sum())
print(matches_with_position["away_position"].isnull().sum())
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=matches_with_position, x="target", y="position_gap", order=["H", "D", "A"])
plt.title("Position Gap (Away Position − Home Position) vs Match Outcome")
plt.xlabel("Match Outcome")
plt.ylabel("Position Gap (positive = home team finished higher)")
plt.axhline(0, color="gray", linestyle="--")
plt.show()
```

median for the position gap of each outcomes-> home (4), draw (-1), away (-6)

```python
print(matches_with_position.columns.tolist())
```

```python
matches_with_position["gap_bucket"] = pd.cut(
    matches_with_position["position_gap"],
    bins=[-20, -15, -10, -5, 0, 5, 10, 15, 20],
    labels=["-20 to -15", "-15 to -10", "-10 to -5", "-5 to 0", "0 to 5", "5 to 10", "10 to 15", "15 to 20"]
)

outcome_by_gap = pd.crosstab(matches_with_position["gap_bucket"], matches_with_position["target"], normalize="index") * 100
outcome_by_gap = outcome_by_gap[["H", "D", "A"]]
print(outcome_by_gap)
```

'Point gap' binned to see more detailed probability of Home or Away side winning.
