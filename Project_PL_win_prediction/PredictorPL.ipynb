```python
matches_clean = matches_clean.drop_duplicates()
print(matches_clean.shape)
```

```python
columns_to_keep = [
    "id", "home", "away", "date", "year", "venue", "league",
    "game_status", "home_score", "away_score",
    "home_possessionPct", "away_possessionPct",
    "home_shotsSummary", "away_shotsSummary",
    "home_foulsCommitted", "away_foulsCommitted",
    "home_yellowCards", "away_yellowCards",
    "home_redCards", "away_redCards",
    "home_wonCorners", "away_wonCorners"
]

matches_clean = matches_clean[columns_to_keep]
matches_clean.head()
```

```python
matches_clean["match_date"] = pd.to_datetime(
    matches_clean["date"] + ", " + 
    np.where(matches_clean["date"].str.contains("January|February|March|April|May|June|July"), 
             (matches_clean["year"] + 1).astype(str), 
             matches_clean["year"].astype(str)),
    format="%A, %B %d, %Y"
)

#Check if weekday in the "date" column matches the actual weekday of the "match_date" column
# weekday = matches_clean["date"].str.split(",").str[0]
# ok = matches_clean["match_date"].dt.day_name() == weekday
# print(ok.all())
# print(matches_clean.loc[~ok, ["date", "year", "match_date"]].head(20))
```

```python
#Cleaned up piece
matches_clean["season_start_year"] = matches_clean["year"]
matches_clean = matches_clean.drop(columns=["date", "year"])
matches_clean[["match_date", "season_start_year", "league"]].head()
```

```python
#Removing % signs and turning them into float
matches_clean["home_possessionPct"] = matches_clean["home_possessionPct"].str.replace("%", "").astype(float)
matches_clean["away_possessionPct"] = matches_clean["away_possessionPct"].str.replace("%", "").astype(float)
```

```python
#Separating the total shots and shots on target into separate columns
matches_clean["home_shots_total"] = matches_clean["home_shotsSummary"].str.extract(r"(\d+)").astype(float)
matches_clean["home_shots_on_target"] = matches_clean["home_shotsSummary"].str.extract(r"\((\d+)\)").astype(float)

matches_clean["away_shots_total"] = matches_clean["away_shotsSummary"].str.extract(r"(\d+)").astype(float)
matches_clean["away_shots_on_target"] = matches_clean["away_shotsSummary"].str.extract(r"\((\d+)\)").astype(float)

matches_clean = matches_clean.drop(columns=["home_shotsSummary", "away_shotsSummary"])
matches_clean[["home_shots_total", "home_shots_on_target", "away_shots_total", "away_shots_on_target"]].tail()
```

```python
matches_clean[["home_shots_total", "home_shots_on_target", "away_shots_total", "away_shots_on_target"]].describe()
```

```python
#turning the early season stats into NaN values since they are not reliable
early_season_mask = matches_clean["season_start_year"] == 2001

stat_columns = [
    "home_possessionPct", "away_possessionPct",
    "home_shots_total", "home_shots_on_target",
    "away_shots_total", "away_shots_on_target",
    "home_foulsCommitted", "away_foulsCommitted",
    "home_yellowCards", "away_yellowCards",
    "home_redCards", "away_redCards",
    "home_wonCorners", "away_wonCorners"
]

matches_clean.loc[early_season_mask, stat_columns] = np.nan
```

```python
#Checking if all goals are numeric, if not converting them
matches_clean["home_score"] = matches_clean["home_score"].astype(int)
matches_clean["away_score"] = matches_clean["away_score"].astype(int)

print(matches_clean["home_score"].describe())
print(matches_clean["away_score"].describe())
```

```python
#Checking for impossible scores
matches_clean[(matches_clean["home_score"] > 10) | (matches_clean["away_score"] > 10)]
```

```python
#how many matches ended in a draw, home win, or away win
def get_result(row):
    if row["home_score"] > row["away_score"]:
        return "H"
    elif row["home_score"] < row["away_score"]:
        return "A"
    else:
        return "D"

matches_clean["target"] = matches_clean.apply(get_result, axis=1)
matches_clean["target"].value_counts()
```

```python
#Cross check against example (home tends to win more often)
matches_clean[["home_score", "away_score", "target"]].head(10)
```

```python
#3 stragglers with 0% possession as found before which need fixing
matches_clean[(matches_clean["home_possessionPct"] == 0) & (matches_clean["season_start_year"] != 2001)]
```

```python
#Fixing the other stragglers that have 0% possession but are not from the 2001-2002 season
stat_columns = [
    "home_possessionPct", "away_possessionPct",
    "home_shots_total", "home_shots_on_target",
    "away_shots_total", "away_shots_on_target",
    "home_foulsCommitted", "away_foulsCommitted",
    "home_yellowCards", "away_yellowCards",
    "home_redCards", "away_redCards",
    "home_wonCorners", "away_wonCorners"
]

stragglers_mask = (matches_clean["home_possessionPct"] == 0) & (matches_clean["season_start_year"] != 2001)
matches_clean.loc[stragglers_mask, stat_columns] = np.nan
```
