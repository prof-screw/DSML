Mapping Type:
1. represented in python by dictionary(key value pair)
2. key value must be primitive datatype
3. Mutable

```python
myDict= {"name": "Bruce Wayne",
         1 : "Batman", 
         "city": {"city1": "Gotham", "city2": "Metropolis"},
         "dictionary": {"key1": "value1", "key2": "value2"}
         }
myDict
```

```python
myDict[1]="superman"  #
myDict
```

```python
myDict.get("dictionary").get("key1")  # returns the value associated with the key "dictionary"
```

```python
myDict["age"]= 35 #add new key value pair like this
myDict
```

```python
myDict.popitem()
```

```python
myDict.pop(1) #remove key value pair by key
```

```python
myDict.keys()
myDict.values()
myDict2= myDict.copy()
```

```python
myDict2
```

```python
myDict2.clear()
```

```python
myDict2.items()
```

```python
dict(sorted(myDict2.items())) #all keys have to be the same data type to sort dictionary
```
