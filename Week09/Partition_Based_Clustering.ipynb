### K means algorithm

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df= pd.read_csv("../Dataset/Iris.csv", index_col= 0)

df.head()
```

```python
features=df.drop("Species", axis=1).to_numpy()
features
```

```python
scaler=StandardScaler()
x=scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans 
```

```python
kmean_algo=KMeans(n_clusters=3,init="random")
```

```python
kmean_algo.fit(x)
```

```python
kmean_algo.labels_
```

```python
import seaborn as sns
```

```python
d=kmean_algo.labels_
sns.scatterplot(data=df,x="Sepal Length (cm)",y='Petal Length (cm)',hue=d)
```
